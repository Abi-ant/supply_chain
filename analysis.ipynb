```python
import duckdb
import pandas as pd
import numpy as np
from scipy import stats

# 1. Connect to DuckDB instance
con = duckdb.connect('supply_chain.duckdb')

# 2. Load order-level summary for complexity and delay analysis
df_orders = con.execute("""
    SELECT 
        order_id,
        order_region,
        market,
        shipping_mode,
        total_line_items,
        total_item_units,
        total_order_value,
        actual_shipping_days,
        scheduled_shipping_days,
        max_lead_time_variance,
        is_late_delivery
    FROM int_order_fulfillment_summary
""").df()

# 3. Load item-level data for inventory velocity analysis
df_items = con.execute("""
    SELECT 
        order_item_id,
        order_id,
        category_name,
        item_quantity,
        sales_amount,
        profit_ratio,
        lead_time_variance_days,
        calculated_delivery_status
    FROM stg_supply_chain
""").df()

print("Loaded DataFrames successfully! Total orders analyzed:", len(df_orders))
```
